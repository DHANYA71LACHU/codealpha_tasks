```python
import pandas as pd

df = pd.read_csv("Protein_Dataset.csv")

df.head()

import matplotlib.pyplot as plt



plt.figure(figsize=(10,5))



df["Organism"].value_counts().plot(kind="bar")



plt.title("Protein Records by Organism")

plt.xlabel("Organism")

plt.ylabel("Number of Proteins")



plt.tight_layout()



plt.savefig("figures/organism_distribution.png", dpi=300)



plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Sequence Length"], bins=20)

plt.title("Protein Sequence Length Distribution")
plt.xlabel("Sequence Length")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    "figures/sequence_length_histogram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
!pip install seaborn
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

sns.boxplot(y=df["Sequence Length"])

plt.title("Sequence Length Boxplot")

plt.tight_layout()

plt.savefig(
    "figures/boxplot_sequence_length.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Sequence Length",
    y="Molecular Weight (Da)"
)

plt.title("Sequence Length vs Molecular Weight")

plt.tight_layout()

plt.savefig(
    "figures/scatter_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
numeric = df.select_dtypes(include="number")

plt.figure(figsize=(6,5))

sns.heatmap(
    numeric.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "figures/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
df["Organism"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)

plt.title("Protein Records by Organism")
plt.ylabel("")

plt.tight_layout()

plt.savefig(
    "figures/pie_chart_organisms.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

df["Organism"].value_counts().plot(
    kind="barh"
)

plt.title("Protein Records by Organism")
plt.xlabel("Number of Proteins")
plt.ylabel("Organism")

plt.tight_layout()

# Save the figure
plt.savefig(
    "figures/organism_distribution_horizontal.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
